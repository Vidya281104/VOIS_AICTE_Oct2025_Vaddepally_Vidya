## End users

- Hosts and property managers: to benchmark prices, optimize listing strategy, and monitor competition.
- Real estate investors: to assess market dynamics, concentration among “power hosts,” and pricing potential by area and room type.
- Revenue/pricing analysts: to design pricing rules, identify outliers, and track performance by neighborhood.
- Marketplaces/OTAs (platform teams): to monitor supply concentration, detect policy risk, and improve search/pricing UX.
- City planners and policymakers: to understand short-term rental density and price impacts across neighborhoods.
- Travel and tourism analysts: to gauge accommodation supply, price ranges, and seasonal patterns.
- Data science and BI teams: to build dashboards, forecasts, and anomaly detection on listing performance.
- Academic/researchers: to study urban economics, tourism effects, and platform market structure.
- Compliance and risk teams: to flag potential multi-listing hosts and policy-sensitive segments.
- Consultants/strategy teams: to translate findings into market-entry, pricing, or operational recommendations.
